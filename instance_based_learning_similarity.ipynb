# Instance-Based Learning (k-NN) with Different Similarity Measures
This notebook demonstrates a simple instance-based learning approach using k-Nearest Neighbors.
We compare results using:
- Euclidean Distance
- Manhattan Distance
- Cosine Similarity


```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Simple dataset: [Height, Weight]
X = np.array([
    [150, 50],
    [160, 60],
    [170, 65],
    [180, 80],
    [175, 75]
])

# Labels: 0 = Small build, 1 = Large build
y = np.array([0, 0, 0, 1, 1])

# Test instance
test = np.array([172, 68])

print("Dataset:\n", X)
print("Labels:\n", y)
print("Test Instance:\n", test)
```

```python
# Euclidean Distance
def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

# Manhattan Distance
def manhattan_distance(a, b):
    return np.sum(np.abs(a - b))

# Cosine Similarity
def cosine_sim(a, b):
    return cosine_similarity([a], [b])[0][0]
```

```python
def knn_predict(X, y, test, k=3, metric='euclidean'):
    distances = []
    
    for i in range(len(X)):
        if metric == 'euclidean':
            dist = euclidean_distance(X[i], test)
        elif metric == 'manhattan':
            dist = manhattan_distance(X[i], test)
        elif metric == 'cosine':
            dist = 1 - cosine_sim(X[i], test)  # convert similarity to distance
        
        distances.append((dist, y[i]))
    
    distances.sort(key=lambda x: x[0])
    neighbors = distances[:k]
    
    labels = [label for _, label in neighbors]
    prediction = max(set(labels), key=labels.count)
    
    return prediction, neighbors
```

```python
metrics = ['euclidean', 'manhattan', 'cosine']

for m in metrics:
    pred, neigh = knn_predict(X, y, test, k=3, metric=m)
    print("\nMetric:", m)
    print("Nearest Neighbors:", neigh)
    print("Prediction:", pred)
```

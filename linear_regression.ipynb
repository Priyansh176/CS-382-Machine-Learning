```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create Dataset
np.random.seed(0)

x = np.linspace(0, 10, 30)
y = 3*x + 5 + np.random.randn(30)*3   # true line: 3x + 5 + noise

plt.scatter(x, y)
plt.title("Data points")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# Prediction Function
def predict(x, b0, b1):
    return b0 + b1*x
```

```python
# Error Function
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)
```

## Brute Force Method

```python
best_error = 1e9
best_b0, best_b1 = 0, 0

for b0 in np.linspace(-10, 10, 50):
    for b1 in np.linspace(-5, 5, 50):
        y_hat = predict(x, b0, b1)
        err = mse(y, y_hat)
        if err < best_error:
            best_error = err
            best_b0, best_b1 = b0, b1

print(best_b0, best_b1, best_error)
```

## Gradient Descent Method

```python
b0, b1 = 0, 0
lr = 0.01
n = len(x)

losses = []

for _ in range(1000):

    y_hat = predict(x, b0, b1)

    db0 = (-2/n) * np.sum(y - y_hat)
    db1 = (-2/n) * np.sum(x*(y - y_hat))

    b0 -= lr * db0
    b1 -= lr * db1

    losses.append(mse(y, y_hat))

print(b0, b1)
```

```python
plt.scatter(x, y)
plt.plot(x, predict(x, b0, b1), color='red')
plt.title("Final fitted regression line")
plt.show()
```

```python
plt.plot(losses)
plt.title("Loss decreasing over iterations")
plt.xlabel("Iteration")
plt.ylabel("MSE")
plt.show()
```

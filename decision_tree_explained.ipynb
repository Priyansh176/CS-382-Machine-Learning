# Decision Tree (Information Gain vs Gini Index)

Dataset used: Classic **Play Tennis Dataset**


```python
import pandas as pd
import math

data = {
    'Outlook': ['Sunny','Sunny','Overcast','Rain','Rain','Rain','Overcast','Sunny','Sunny','Rain','Sunny','Overcast','Overcast','Rain'],
    'Temperature': ['Hot','Hot','Hot','Mild','Cool','Cool','Mild','Cool','Mild','Mild','Mild','Mild','Hot','Mild'],
    'Humidity': ['High','High','High','High','Normal','Normal','Normal','High','Normal','Normal','Normal','High','Normal','High'],
    'Wind': ['Weak','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Strong'],
    'PlayTennis': ['No','No','Yes','Yes','Yes','No','Yes','No','Yes','Yes','Yes','Yes','Yes','No']
}

df = pd.DataFrame(data)
```

## Step 1: Entropy of Target

Entropy formula:

Entropy(S) = - Σ p(x) log2 p(x)


```python
def entropy(column):
    values = column.value_counts(normalize=True)
    return -sum(p * math.log2(p) for p in values)

entropy(df['PlayTennis'])
```

## Step 2: Information Gain

Information Gain = Entropy(parent) - Weighted Entropy(children)


```python
def info_gain(df, attr, target='PlayTennis'):
    total_entropy = entropy(df[target])
    values = df[attr].unique()
    
    weighted_entropy = 0
    for v in values:
        subset = df[df[attr] == v]
        weighted_entropy += (len(subset)/len(df)) * entropy(subset[target])
    
    return total_entropy - weighted_entropy

for col in df.columns[:-1]:
    print(col, info_gain(df, col))
```

Attribute with highest Information Gain becomes ROOT.
Continue splitting recursively.

(This is how decision tree using entropy works)


# Now Using Gini Index

Gini formula:

Gini = 1 - Σ (p(x))²


```python
def gini(column):
    values = column.value_counts(normalize=True)
    return 1 - sum(p**2 for p in values)

def gini_split(df, attr, target='PlayTennis'):
    values = df[attr].unique()
    weighted = 0
    for v in values:
        subset = df[df[attr] == v]
        weighted += (len(subset)/len(df)) * gini(subset[target])
    return weighted

for col in df.columns[:-1]:
    print(col, gini_split(df, col))
```

Attribute with LOWEST Gini becomes ROOT.

This is how CART decision tree works.
